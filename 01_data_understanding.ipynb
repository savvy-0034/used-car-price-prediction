## Initial Findings – Craigslist Vehicles Dataset

- The Craigslist dataset contains 21,972 vehicle records and 26 columns.
- The first 1,000 records were inspected for initial data understanding.
- The dataset contains both numerical and categorical vehicle-related attributes.
- The `description` column contains textual vehicle listing information and will be used for the later NLP/TF-IDF stage.
- Among the first 1,000 records, 27 records had missing vehicle descriptions.
- Missing values were observed in several vehicle attributes. The highest missing-value percentages were found in condition (42.0%), cylinders (34.7%), drive (31.3%), and type (29.3%).
- The `price` column was available for all 1,000 inspected records.
- The dataset contains vehicle information such as manufacturer, model, fuel type, mileage, transmission, drivetrain, vehicle type, and location.
- No data cleaning or modification was performed at this stage; the original dataset was retained for further processing.
- The most common vehicle types in the inspected sample were sedan, SUV, truck, and pickup.
- Among vehicles with available condition information, good-condition vehicles were the most common.

```python
# Check missing values in the Craigslist dataset

print("Missing values in each column:")
print(craigslist_sample[relevant_columns].isnull().sum())
```

```python
# Calculate missing value percentages

missing_percentage = (
    craigslist_sample[relevant_columns].isnull().sum()
    / len(craigslist_sample) * 100
)

print("Missing value percentages:")
print(missing_percentage.round(2))
```

```python
print("Highest missing-value columns:")
print(missing_percentage.sort_values(ascending=False).head(5))
```
