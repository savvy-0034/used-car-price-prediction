```python
pca_features = [
    'Present_Price',
    'Kms_Driven',
    'Owner',
    'Car_Age',
    'Fuel_Type_Diesel',
    'Fuel_Type_Petrol',
    'Seller_Type_Individual',
    'Transmission_Manual'
]

X_pca = df_encoded[pca_features]

X_pca.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_pca)

print(X_scaled.shape)
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA()

X_pca_result = pca.fit_transform(X_scaled)

print("Original features:", X_scaled.shape[1])
print("PCA components:", X_pca_result.shape[1])
```

```python
explained_variance = pca.explained_variance_ratio_

for i, variance in enumerate(explained_variance, start=1):
    print(f"PC{i}: {variance:.4f} ({variance*100:.2f}%)")
```

```python
cumulative_variance = np.cumsum(pca.explained_variance_ratio_)

for i, variance in enumerate(cumulative_variance, start=1):
    print(f"PC1-PC{i}: {variance:.4f} ({variance*100:.2f}%)")
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker='o'
)

plt.axhline(y=0.90, linestyle='--', label='90% variance')
plt.axhline(y=0.95, linestyle='--', label='95% variance')

plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('PCA Cumulative Explained Variance')

plt.legend()
plt.show()
```

```python
explained_variance = pca.explained_variance_ratio_

for i, variance in enumerate(explained_variance, start=1):
    print(f"PC{i}: {variance:.4f} ({variance*100:.2f}%)")
```

```python
cumulative_variance = np.cumsum(pca.explained_variance_ratio_)

for i, variance in enumerate(cumulative_variance, start=1):
    print(f"PC1-PC{i}: {variance:.4f} ({variance*100:.2f}%)")
```

```python
pca_6 = PCA(n_components=6)

X_pca_6 = pca_6.fit_transform(X_scaled)

print("Original feature count:", X_scaled.shape[1])
print("Reduced feature count:", X_pca_6.shape[1])
```

```python
print(X_pca_6[:5])
```

```python
pca_columns = [f'PC{i}' for i in range(1, 7)]

X_pca_df = pd.DataFrame(
    X_pca_6,
    columns=pca_columns,
    index=df.index
)

X_pca_df.head()
```

```python
print("PCA dataset shape:", X_pca_df.shape)
print("Total variance retained:",
      pca_6.explained_variance_ratio_.sum())
```

```python
for i, variance in enumerate(
    pca_6.explained_variance_ratio_,
    start=1
):
    print(f"PC{i}: {variance*100:.2f}%")
```

```python
final_pca_df = X_pca_df.copy()

final_pca_df['Selling_Price'] = df['Selling_Price'].values

final_pca_df.head()
```
